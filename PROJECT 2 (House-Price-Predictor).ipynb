```python
def convertRange(x):
    if isinstance(x,str):
        temp = x.split('-')
        if len(temp) == 2:
            return (float(temp[0].strip()) + float(temp[1].strip())) / 2  
    try:
        return float(x)
    except:
        return None
    
```

```python
def remove_outliers_sqft(df):
    df_output = pd.DataFrame()
    for key, subdf in df.groupby('location'):
        m = np.mean(subdf.price_per_sqft)

        st = np.std(subdf.price_per_sqft)

        gen_df = subdf[(subdf.price_per_sqft > (m-st)) & (subdf.price_per_sqft <= (m+st))]
        df_output = pd.concat([df_output,gen_df], ignore_index=True)
    return df_output
df = remove_outliers_sqft(df)
df.describe()
```

```python
def BHK_outliers_remover(df):
    exclude_indices = np.array([])
    for location, location_df in df.groupby('location'):
        BHK_stats = {}
        for BHK, BHK_df in location_df.groupby('BHK'):
            BHK_stats[BHK] = {
                'mean' : np.mean(BHK_df.price_per_sqft),
                'std' : np.std(BHK_df.price_per_sqft),
                'count' : BHK_df.shape[0]
            }
        for BHK, BHK_df in location_df.groupby('BHK'):
            stats = BHK_stats.get(BHK-1)
            if stats and stats['count'] > 5:
                exclude_indices = np.append(exclude_indices,
                                            BHK_df[BHK_df.price_per_sqft<(stats['mean'])
                                                ].index.values)
    return df.drop(exclude_indices, axis='index')                                
```

```python
df = BHK_outliers_remover(df)
```

```python
x = df.drop(columns=['price'])
y = df['price']
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression,Lasso,Ridge
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import make_column_transformer
from sklearn.pipeline import make_pipeline
from sklearn.metrics import r2_score
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,
                                                 test_size=0.2,
                                                 random_state=0)
```

```python
x_train.shape,x_test.shape
```

### Apply Linear Regression :

```python
column_trans = make_column_transformer((OneHotEncoder(sparse_output=False), ['location']),
                                        remainder='passthrough')
```

```python
scaler = StandardScaler()
```

```python
lr = LinearRegression()
```

```python
pipe = make_pipeline(column_trans, scaler, lr)
```

```python
pipe.fit(x_train,y_train)
```

```python
y_pred_lr = pipe.predict(x_test)
r2_score(y_test,y_pred_lr)
```

### Applying Lasso

```python
lasso = Lasso()
```

```python
pipe = make_pipeline(column_trans, scaler, lasso)
```

```python
pipe.fit(x_train,y_train)
```

```python
y_pred_lasso = pipe.predict(x_test)
r2_score(y_test,y_pred_lasso)
```

### Applying Ridge

```python
ridge = Ridge()
```

```python
pipe = make_pipeline(column_trans, scaler, ridge)
```

```python
pipe.fit(x_train,y_train)
```

```python
y_pred_ridge = pipe.predict(x_test)
r2_score(y_test,y_pred_ridge)
```

```python
print("No Regularization: ",r2_score(y_test,y_pred_lr))
print("Lasso:             ",r2_score(y_test,y_pred_lasso))
print("Ridge:             ",r2_score(y_test,y_pred_ridge))      
```

```python
for a in [ 0.01, 0.1, 1,30,50,500,550,560]:
    ridge_test = Ridge(alpha=a)
    pipe = make_pipeline(column_trans, scaler, ridge_test)
    pipe.fit(x_train,y_train)

    y_pred_test = pipe.predict(x_test)
    score = r2_score(y_test,y_pred_test)
    print(a, "--->", score)
```

```python
ridge_final = Ridge(alpha=550)
pipe = make_pipeline(column_trans, scaler, ridge_final)
pipe.fit(x_train,y_train)
y_pred_final = pipe.predict(x_test)
print("Final r2_score: ",r2_score(y_test,y_pred_final))
```
